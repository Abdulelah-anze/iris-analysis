# part 4: Export Results

```python
import os
os.makedirs("output", exist_ok=True)

species_summary = df.groupby('species').mean(numeric_only=True)
species_summary.to_csv("output/species_summary.csv")

filtered = df[df['sepal_length'] > 6]
filtered.to_csv("output/filtered_iris.csv", index=False)
```

# part5: Additional Analysis and Visualizations

```python
df["sepal_length"].plot.hist()
```

The histogram shows the distribution of sepal length values in the dataset. Most flowers have sepal lengths between about 5 and 6.5, while fewer flowers have very short or very long sepals.

```python
df.plot.scatter(x="sepal_length", y="petal_length")
```

The scatter plot shows a positive relationship between sepal length and petal length. In general, flowers with longer sepals tend to also have longer petals.

```python
df.boxplot(column="petal_width", by="species")
```

The box plot compares petal width across the three Iris species. Setosa has the smallest petal width, versicolor is in the middle, and virginica has the largest petal width.

```python
avg_petal_width = df.groupby("species")["petal_width"].mean()

avg_petal_width.plot.bar()
```

The bar chart compares the average petal width for each species. Virginica has the highest average petal width, followed by versicolor, while setosa has the lowest average.

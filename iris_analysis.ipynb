# part5: Additional Analysis and Visualizations

```python
avg_sepal_width = df.groupby("species")["sepal_width"].mean()

avg_sepal_width.plot.barh()
```

This chart compares the average sepal width among the three Iris species. Setosa has the highest average sepal width, while versicolor has the lowest.

```python
long_petals = df[df["petal_length"] > 4]

long_petals["species"].value_counts().plot.bar()
```

This chart shows the number of flowers with a petal length greater than 4 for each species. It helps us compare which species contains more flowers with long petals.

```python
avg_sepal_width = df.groupby("species")["sepal_width"].mean()

avg_sepal_width.plot.pie(
    autopct="%1.1f%%",
    ylabel="",
    title="Average Sepal Width by Species"
)
```

This pie chart compares the average sepal width among the three Iris species. Setosa has the highest average sepal width, while versicolor has the lowest.

The scatter plot shows a positive relationship between sepal length and petal length. In general, flowers with longer sepals tend to also have longer petals.

```python
df.boxplot(column="petal_width", by="species")
```

The box plot compares petal width across the three Iris species. Setosa has the smallest petal width, versicolor is in the middle, and virginica has the largest petal width.
